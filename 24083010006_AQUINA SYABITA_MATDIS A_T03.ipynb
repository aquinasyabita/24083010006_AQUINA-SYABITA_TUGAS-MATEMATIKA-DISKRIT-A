```python
# ==========================
# SOAL 1
# n^2 - n selalu habis dibagi (n-1)
# ==========================

def cek_pembagian(n_max=15):
    print("SOAL 1: Cek apakah n^2 - n habis dibagi (n-1)\n")
    for n in range(2, n_max+1):  # mulai dari 2 supaya tidak bagi nol
        hasil = (n**2 - n) % (n-1) == 0
        print(f"n={n:2d} -> {n**2 - n} / {n-1} habis dibagi? {hasil}")

cek_pembagian(15)

print("\nKESIMPULAN SOAL 1:")
print("Karena (n^2 - n) = n(n-1), maka pasti habis dibagi (n-1) untuk semua n > 1.")
```

```python
# ==========================
# SOAL 2
# Cari n minimum sehingga H_n >= 2
# ==========================

def cari_n_batas(target=2):
    total = 0.0
    n = 0
    while total < target:
        n += 1
        total += 1/n
    return n, total

n_batas, jumlah = cari_n_batas(2)

print("SOAL 2: Deret harmonik\n")
print(f"n minimum = {n_batas}")
print(f"H_{n_batas} = {jumlah}")

print("\nKESIMPULAN SOAL 2:")
print(f"Deret harmonik pertama kali mencapai atau melebihi 2 pada n = {n_batas},")
print(f"dengan jumlah H_{n_batas} = {jumlah:.6f}.")
```
