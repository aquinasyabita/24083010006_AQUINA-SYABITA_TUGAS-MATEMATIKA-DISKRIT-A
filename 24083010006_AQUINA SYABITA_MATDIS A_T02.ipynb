```python
def odd_sum(n):
    return sum(2*i - 1 for i in range(1, n+1))

print("SOAL 1: 1 + 3 + 5 + ... + (2n-1) = n^2")

# Basis Induksi
n = 1
jumlah = odd_sum(n)
rumus = n**2
status = (jumlah == rumus)
print(f"Basis Induksi n={n}: jumlah deret = {jumlah}, n^2 = {rumus} -> Status: {status}")

# Cek beberapa nilai n
for k in range(2, 6):
    jumlah = odd_sum(k)
    rumus = k**2
    status = (jumlah == rumus)
    print(f"n={k}: jumlah deret = {jumlah}, n^2 = {rumus} -> Status: {status}")

print("\nKESIMPULAN: 1 + 3 + 5 + ... + (2n-1) = n^2 untuk semua n >= 1")
```

```python
def sum_squares(n):
    return sum(i**2 for i in range(1, n+1))

print("SOAL 3: 1^2 + 2^2 + 3^2 + ... + n^2 = n(n+1)(2n+1)/6")

# Basis Induksi
n = 0
jumlah = sum_squares(n)
rumus = n*(n+1)*(2*n+1)//6
status = (jumlah == rumus)
print(f"Basis Induksi n={n}: jumlah deret = {jumlah}, n(n+1)(2n+1)/6 = {rumus} -> Status: {status}")

# Cek beberapa nilai n
for k in range(1, 6):
    jumlah = sum_squares(k)
    rumus = k*(k+1)*(2*k+1)//6
    status = (jumlah == rumus)
    print(f"n={k}: jumlah deret = {jumlah}, n(n+1)(2n+1)/6 = {rumus} -> Status: {status}")

print("\nKESIMPULAN: 1^2 + 2^2 + ... + n^2 = n(n+1)(2n+1)/6 untuk semua n >= 0")
```
